# <center>**NumPy**<center>

- knihovna Numerical Python, přidává matematické funkce a operace
- jako každou knihovnu je třeba ji mít nejprve nainstalovanou
- po instalaci se knihovny do kódu přidává pomocí příkazu `import [jméno knihovny malými písmeny]`
- v základu bychom Numpy tedy importovali jako `import numpy`
- funkce se následně volají ve tvaru `[název knihovny].[název funkce](argument)`
- přidáním příkazu `as` za název si můžeme název knihovny přejmenovat (nejčastěji se použije nějaká zkratka)
- pro Numpy standard: `import numpy as np`
- funkce se pak volají takto: `[zkratka knihovny].[název funkce](argument)`

```python
import numpy as np # import knihovny + zavedení zkratky

# NumPy arrays
F = np.array(seznam)
print(F) # NumPy array vše hází na stejný formát
print(seznam) # klasický list z Pythonu v sobě může mít více různých hodnot

# Konstanty, logaritmy + goniometrické funkce
print(np.log(np.e))
print(np.sin(np.pi))
```

# <center>**Pandas**<center>

- knihovna pro práci s tabulkami
- bere .txt i .xlsx (Excel) soubory
- pamatovat: Python indexuje od 0
- u texťáků potřeba dávat při načítání pozor na separátory (někdy si to neumí vybrat správný)

```python
# Načtení knihovny
import pandas as pd

# Načtení tabulky z Excelu
rf = pd.read_excel('rf_data.xlsx')
print(rf)
print()
```

```python
# Přístup k jednotlivým hodnotám
print(f'Hodnota  sloupce N v 2. řádku je {rf.at[1, 'N']}')
print()
print(f'Hodnoty proměnných v druhém řádku jsou:')
print(f'{rf.iloc[1]}')
```

```python
# Nahrazení hodnot
rf.at[2,'t'] = 666
rf
```

```python
# Poslední hodnota 
print(f'Poslední hodnota ve sloupci f je {rf['f'].iat[-1]}') # index -1 značí poslední hodnotu ze seznamu
print()
print(f'Poslední hodnoty všech sloupců jsou:')
print(f'{rf.iloc[-1]}')
```
